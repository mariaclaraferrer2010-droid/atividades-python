4. Par ou ímpar

```python
numero = int(input("Digite um número inteiro: "))

if numero % 2 == 0:
    print("Par")
else:
    print("Ímpar")
```

5. Situação do aluno

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")

print(f"Média: {media:.2f}")
```

6. Maior entre três números

```python
num1 = int(input("Primeiro número: "))
num2 = int(input("Segundo número: "))
num3 = int(input("Terceiro número: "))

maior = num1

if num2 > maior:
    maior = num2
if num3 > maior:
    maior = num3

print("Maior número:", maior)
```

7. Contagem crescente

```python
n = int(input("Digite um número inteiro positivo N: "))

for i in range(1, n + 1):
    print(i)
```

8. Somando de 1 até N

```python
n = int(input("Digite N: "))

soma = 0
for i in range(1, n + 1):
    soma += i

print("Soma:", soma)
```

9. Contando números pares

```python
pares = 0

for i in range(10):
    numero = int(input("Digite um número inteiro: "))
    if numero % 2 == 0:
        pares += 1

print(f"Foram digitados {pares} números pares.")
```

10. Soma apenas dos positivos

```python
soma_positivos = 0

for i in range(8):
    numero = int(input("Digite um número inteiro: "))
    if numero > 0:
        soma_positivos += numero

print("Soma dos positivos:", soma_positivos)
```

11. Maior valor digitado

```python
maior = int(input("Digite o 1º número: "))

for i in range(2, 7):
    numero = int(input(f"Digite o {i}º número: "))
    if numero > maior:
        maior = numero

print("Maior valor:", maior)
```

12. Tabuada

```python
numero = int(input("Digite um número para ver sua tabuada: "))

for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

13. Tentativas de senha

```python
senha = input("Digite a senha: ")

while senha != "1234":
    senha = input("Senha incorreta. Tente novamente: ")

print("Acesso permitido!")
```

14. Número secreto

```python
numero_secreto = 7
tentativas = 0

while True:
    chute = int(input("Tente adivinhar o número secreto: "))
    tentativas += 1

    if chute < numero_secreto:
        print("Tente um número maior")
    elif chute > numero_secreto:
        print("Tente um número menor")
    else:
        print("Parabéns! Você acertou.")
        print(f"Tentativas necessárias: {tentativas}")
        break
```

15. Analisando uma sequência de números

```python
quantidade = 0
positivos = 0
negativos = 0
soma = 0

while True:
    numero = int(input("Digite um número: "))
    if numero == 0:
        break

    quantidade += 1
    soma += numero

    if numero > 0:
        positivos += 1
    elif numero < 0:
        negativos += 1

print("Quantidade:", quantidade)
print("Positivos:", positivos)
print("Negativos:", negativos)
print("Soma:", soma)
```
